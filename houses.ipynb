```python
import numpy as np
import pandas as pd

np.random.seed(42)
square_meters=np.random.randint(low=50, high=200, size=100)
rooms=np.random.randint(low=1, high=5, size=100)

noise=np.random.normal(loc=0, scale=500000, size=100)

price=np.round(square_meters*40000+rooms*150000+noise)


data={
    'square_meters':square_meters,
    'rooms':rooms,
    'price':price
}
df=pd.DataFrame(data)
```

```python
df.to_csv("house",index=False)
```
